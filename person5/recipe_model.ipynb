```python
ingredient_keywords = {
    'has_pork':       ['หมูสับ', 'หมู'],
    'has_chicken':    ['อกไก่', 'ไก่'],
    'has_shrimp':     ['กุ้ง'],
    'has_bacon':      ['เบคอน', 'แฮม'],
    'has_egg':        ['ไข่'],
    'has_rice':       ['ข้าว'],
    'has_pasta':      ['พาสต้า', 'สปาเก็ต', 'เพนเน่', 'ลาซานญ่า'],
    'has_bread':      ['ขนมปัง'],
    'has_holy_basil': ['กะเพรา'],
    'has_tomato':     ['มะเขือเทศ'],
    'has_cheese':     ['ชีส'],
    'has_cabbage':    ['ผักกาด']
}
```

```python
for col_name, keywords in ingredient_keywords.items():
    df[col_name] = df['main_ingredients'].apply(
        lambda x: 1 if any(
            kw in item
            for item in x.split(';')
            for kw in keywords
            if not (item.strip().startswith('ไข่') and kw == 'ไก่')
        ) else 0
    )
```

```python
print(df[list(ingredient_keywords.keys())].sum())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['main_ingredients'])
print(tfidf_matrix.shape)
```

```python
def recommend_tfidf(user_ingredients, top_n=5):
    user_str = ';'.join(user_ingredients)
    user_vector = vectorizer.transform([user_str])
    scores = cosine_similarity(user_vector, tfidf_matrix).flatten()
    top_indices = scores.argsort()[::-1][:top_n]
    result = df[['name_th', 'name_en', 'cuisine']].iloc[top_indices].copy()
    result['score'] = scores[top_indices].round(3)
    return result
```

```python
recommend_tfidf(['หมูสับ', 'ไข่', 'กะเพรา'])
```

```python
from sklearn.neighbors import NearestNeighbors

X = df[list(ingredient_keywords.keys())].values

knn_model = NearestNeighbors(n_neighbors=5, metric='cosine')
knn_model.fit(X)
```

```python
def recommend_knn(user_ingredients, top_n=5):
    user_vector = [
        1 if any(kw in ing for ing in user_ingredients for kw in keywords)
        else 0
        for keywords in ingredient_keywords.values()
    ]
    distances, indices = knn_model.kneighbors([user_vector], n_neighbors=top_n)
    result = df[['name_th', 'name_en', 'cuisine']].iloc[indices[0]].copy()
    result['distance'] = distances[0].round(3)
    return result
```

```python
recommend_knn(['หมูสับ', 'ไข่', 'กะเพรา'])
```

```python
X_train = []
y_train = []

for i, row in df.iterrows():
    recipe_vector = row[list(ingredient_keywords.keys())].values
    ingredient_indices = np.where(recipe_vector == 1)[0]
    
    if len(ingredient_indices) == 0:
        continue
        
    for _ in range(15):
        n = np.random.randint(1, len(ingredient_indices) + 1)
        subset = np.random.choice(ingredient_indices, n, replace=False)
        user_vector = np.zeros(12)
        user_vector[subset] = 1
        X_train.append(user_vector)
        y_train.append(i)

X_train = np.array(X_train)
y_train = np.array(y_train)
print(X_train.shape, y_train.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
```

```python
def recommend_rf(user_ingredients, top_n=5):
    user_vector = [
        1 if any(kw in ing for ing in user_ingredients for kw in keywords)
        else 0
        for keywords in ingredient_keywords.values()
    ]
    proba = rf_model.predict_proba([user_vector])[0]
    top_indices = proba.argsort()[::-1][:top_n]
    result = df[['name_th', 'name_en', 'cuisine']].iloc[top_indices].copy()
    result['proba'] = proba[top_indices].round(3)
    return result
```

```python
recommend_rf(['หมูสับ', 'ไข่', 'กะเพรา'])
```

```python
def calculate_match(user_ingredients, recipe_index):
    
    recipe_vector = df[list(ingredient_keywords.keys())].iloc[recipe_index]
    recipe_needs = [col for col, val in recipe_vector.items() if val == 1]
    
    col_to_name = {
        'has_pork': 'หมู', 'has_chicken': 'ไก่',
        'has_shrimp': 'กุ้ง', 'has_bacon': 'เบคอน',
        'has_egg': 'ไข่', 'has_rice': 'ข้าว',
        'has_pasta': 'พาสต้า', 'has_bread': 'ขนมปัง',
        'has_holy_basil': 'กะเพรา', 'has_tomato': 'มะเขือเทศ',
        'has_cheese': 'ชีส', 'has_cabbage': 'ผักกาดขาว'
    }
    
    user_has = [col for col, keywords in ingredient_keywords.items()
                if any(kw in ing for ing in user_ingredients for kw in keywords)]
    
    have = [col_to_name[c] for c in recipe_needs if c in user_has]
    missing = [col_to_name[c] for c in recipe_needs if c not in user_has]
    match_pct = len(have) / len(recipe_needs) * 100 if recipe_needs else 0
    
    return round(match_pct, 1), have, missing
```

```python
pct, have, missing = calculate_match(['หมูสับ', 'ไข่', 'กะเพรา'], 0)
print(f"Match: {pct}%")
print(f"มีแล้ว: {have}")
print(f"ขาด: {missing}")
```

Test Case

```python
test_cases = [
    {'ingredients': ['หมูสับ', 'ไข่', 'กะเพรา'],                              'expected_id': 'R001'},
    {'ingredients': ['เบคอน', 'ไข่', 'พาสต้า', 'ชีส'],                        'expected_id': 'R018'},
    {'ingredients': ['กุ้ง', 'ไข่'],                                           'expected_id': 'R017'},
    {'ingredients': ['อกไก่', 'ข้าว'],                                         'expected_id': 'R021'},
    {'ingredients': ['พาสต้า', 'มะเขือเทศ'],                                   'expected_id': 'R020'},
    {'ingredients': ['ขนมปัง', 'ชีส'],                                         'expected_id': 'R028'},
    {'ingredients': ['หมูสับ', 'ไข่', 'ผักกาด'],                              'expected_id': 'R035'},
    {'ingredients': ['อกไก่', 'เบคอน', 'ไข่', 'ชีส', 'มะเขือเทศ', 'ผักกาด'], 'expected_id': 'R040'},
    {'ingredients': ['กุ้ง', 'ข้าว'],                                          'expected_id': 'R026'},
    {'ingredients': ['พาสต้า', 'มะเขือเทศ', 'ชีส'],                           'expected_id': 'R024'},
]
```

```python
import time

results = []

for tc in test_cases:
    ings = tc['ingredients']
    exp  = tc['expected_id']

    t0 = time.time()
    r_tfidf = recommend_tfidf(ings, top_n=3)
    ms_tfidf = round((time.time() - t0) * 1000, 2)

    t0 = time.time()
    r_knn = recommend_knn(ings, top_n=3)
    ms_knn = round((time.time() - t0) * 1000, 2)

    t0 = time.time()
    r_rf = recommend_rf(ings, top_n=3)
    ms_rf = round((time.time() - t0) * 1000, 2)

    top1_tfidf = df.loc[r_tfidf.index[0], 'recipe_id']
    top3_tfidf = df.loc[r_tfidf.index, 'recipe_id'].tolist()
    top1_knn   = df.loc[r_knn.index[0],   'recipe_id']
    top3_knn   = df.loc[r_knn.index,   'recipe_id'].tolist()
    top1_rf    = df.loc[r_rf.index[0],    'recipe_id']
    top3_rf    = df.loc[r_rf.index,    'recipe_id'].tolist()

    results.append({
        'test': ', '.join(ings[:3]) + ('...' if len(ings) > 3 else ''),
        'expected': exp,
        'tfidf_top1': top1_tfidf, 'tfidf_hit1': top1_tfidf == exp, 'tfidf_hit3': exp in top3_tfidf,
        'knn_top1':   top1_knn,   'knn_hit1':   top1_knn   == exp, 'knn_hit3':   exp in top3_knn,
        'rf_top1':    top1_rf,    'rf_hit1':    top1_rf    == exp, 'rf_hit3':    exp in top3_rf,
        'ms_tfidf': ms_tfidf, 'ms_knn': ms_knn, 'ms_rf': ms_rf,
    })

df_res = pd.DataFrame(results)
print(df_res[['test','expected','tfidf_top1','knn_top1','rf_top1']].to_string())
```

```python
summary = pd.DataFrame({
    'Algorithm':      ['TF-IDF + Cosine', 'KNN (cosine)', 'Random Forest'],
    'Top-1 Accuracy': [f"{df_res['tfidf_hit1'].mean()*100:.0f}%",
                       f"{df_res['knn_hit1'].mean()*100:.0f}%",
                       f"{df_res['rf_hit1'].mean()*100:.0f}%"],
    'Top-3 Accuracy': [f"{df_res['tfidf_hit3'].mean()*100:.0f}%",
                       f"{df_res['knn_hit3'].mean()*100:.0f}%",
                       f"{df_res['rf_hit3'].mean()*100:.0f}%"],
    'Avg Time (ms)':  [f"{df_res['ms_tfidf'].mean():.2f}",
                       f"{df_res['ms_knn'].mean():.2f}",
                       f"{df_res['ms_rf'].mean():.2f}"],
})
print(summary.to_string(index=False))
```
